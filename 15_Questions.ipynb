# Questions
## 1. Why do we use EDA (Exploratory Data Analysis)?

Answer:
EDA is used to understand the dataset by finding patterns, missing values, outliers, relationships, and trends before building a machine learning model.



## 2. What type of problems are we solving in AI/ML?

Answer:

Classification – Predict categories (Spam/Not Spam)
Regression – Predict continuous values (House Price)
Clustering – Group similar data
Recommendation – Suggest products or movies
Forecasting – Predict future values

## 3. What is an ROC Curve? How is it measured?

Answer:
ROC (Receiver Operating Characteristic) Curve shows the performance of a classification model by plotting:

True Positive Rate (TPR)
False Positive Rate (FPR)

It is measured using AUC (Area Under Curve).

AUC = 1 → Perfect model
AUC = 0.5 → Random prediction

## 4. Why do we use Univariate, Bivariate, and Multivariate Analysis?

Answer:

Univariate: Analyze one variable.
Bivariate: Analyze relationship between two variables.
Multivariate: Analyze relationships among multiple variables.

## 5. What is Standard Deviation?

Answer:
Standard deviation measures how much data values are spread from the mean.

Low SD → Data is close to the mean.
High SD → Data is widely spread.

## 6. What is Measure of Central Tendency?

Answer:
It represents the center of the data.
Types:

Mean
Median
Mode

## 7. What are Covariance and Variance?

Variance:
Measures how much one variable varies from its mean.

Covariance:
Measures how two variables change together.

Positive → Move together
Negative → Move opposite



## 8. What is a Regression Matrix (Metrics)?

Answer:
Regression metrics evaluate regression models.
Common metrics:

MAE
MSE
RMSE
R² Score

## 9. What is a Classification Matrix?

Answer:
It is the Confusion Matrix, which contains:

True Positive (TP)
True Negative (TN)
False Positive (FP)
False Negative (FN)

Used to calculate Accuracy, Precision, Recall, and F1-score.

## 10. How to calculate Learning Rate?

Answer:
Learning rate is a hyperparameter that controls how much model weights are updated in each iteration.

Weight Update Formula:

New Weight=Old Weight−(Learning Rate×Gradient)

Common values: 0.1, 0.01, 0.001

## 11. What are Overfitting and Underfitting?

Overfitting:
Model learns training data too well and performs poorly on new data.

Underfitting:
Model is too simple and performs poorly on both training and testing data.

## 12. What is Encoding and Decoding?

Encoding:
Converts categorical/text data into numerical form.
Example: Male → 0, Female → 1

Decoding:
Converts encoded values back to their original form.

## 13. What are Forward Propagation and Backward Propagation?

Forward Propagation:
Input passes through the neural network to produce predictions.

Backward Propagation:
Calculates errors and updates weights to reduce the loss.

## 14. Difference between a Perceptron and a Neuron
### Perceptron:
  Single-layer model
  Uses step activation
  Solves linear problems
### Neuron:
   Basic unit of neural networks
   Can use ReLU, Sigmoid, Tanh, etc.
   Solves linear and non-linear problems

## 15. What are Scalar, Vector, Matrix, and Tensor?

Scalar: Single value (5)

Vector: 1D array ([1, 2, 3])

Matrix: 2D array

[[1,2],
 [3,4]]

Tensor: Multi-dimensional array (3D or higher), used in deep learning
