```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style 
```

```python
classes=['python','R','AI','ML','DS']
class1_St=[30,10,20,25,10]
class2_St=[4,5,20,20,10]
class3_St=[3,4,20,10,30]
```

```python
plt.bar(classes,class1_St,color='y')
plt.bar(classes,class2_St,color='m')
plt.bar(classes,class3_St,color='r')
```

```python
plt.bar(classes,class1_St,width=0.6,align='edge',color='m',edgecolor ='r',linewidth=2,alpha=0.5,linestyle='-.')
```

```python
plt.bar(classes,class1_St,width=0.6,align='edge',color='m',edgecolor ='r',linewidth=2,alpha=0.9,linestyle='-.',label='class1students',visible=True)
plt.legend()
```

```python
style.use('ggplot')
plt.bar(classes,class1_St,width=0.6,align='edge',color='m',edgecolor ='r',linewidth=2,alpha=0.9,linestyle='-.',label='class1students',visible=True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.bar(classes,class1_St,width=0.4,align='edge',color='m',edgecolor ='r',linewidth=2,alpha=0.9,linestyle='-.')
plt.title("Bar Chart of MU students",fontsize=15)
plt.xlabel('classes',fontsize=16)
plt.ylabel('No of students',fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(10,10))
plt.bar(classes,class1_St,width=0.4,color='r',label ='Class1 Student')
plt.bar(classes,class2_St,width=0.2,color='m',label ='Class2 Student')
plt.title("Bar Chart of MU students",fontsize=15)
plt.xlabel('classes',fontsize=16)
plt.ylabel('No of students',fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(10,10))
classes_index=np.arange(len(classes))
width=0.2
plt.bar(classes_index,class1_St,width,color='r',label ='Class1 Student')
plt.bar(classes_index + width,class2_St,width,color='m',label ='Class2 Student')
plt.title("Bar Chart of MU students",fontsize=15)
plt.xlabel('classes',fontsize=16)
plt.ylabel('No of students',fontsize=16)
plt.show()
```
